```python
season_demand = df.groupby("season")["cnt"].mean()
print(f"season demand : {season_demand}")
```

```python
season_user = df.groupby("season")[["casual", "registered"]].mean()
print(f"season user : {season_user}")
```

```python
season_weather = df.groupby("season")[["temp", "atemp", "hum", "windspeed"]].mean()
print(f"season weather : {season_weather}")
```

Season 3 has the highest rental demand while Season 1 has the lowest; registered users rent more bikes than casual users, and seasonal weather patterns can help the ML model predict rental demand.

```python
season_weather_count = pd.crosstab(
    df["season"],
    df["weathersit"]
)

print(f"season weather count : {season_weather_count}")
```

```python
season_weather_pct = pd.crosstab(
    df["season"],
    df["weathersit"],
    normalize="index"
) * 100
print(f"season weather percentage : {season_weather_pct}")
```

Fall has the highest clear weather percentage, while Winter has the highest mist percentage; weather distribution can help the ML model learn patterns related to bike rental demand.

```python
weather_rental = df.groupby("weathersit")["cnt"].mean()
print(f"weather rental : {weather_rental}")
```

```python
weekday_rental = df.groupby("weekday")["cnt"].mean()
print(f"weekday rental : {weekday_rental}")
```

```python
hourly_rental = df.groupby("hr")["cnt"].mean()
print(f"hourly rental : {hourly_rental}")
```

```python
workingday_rental = df.groupby("workingday")["cnt"].mean()
print(f"workingday rental : {workingday_rental}")
```

```python
workingday_hour_rental = df.groupby(
    ["workingday", "hr"]
)["cnt"].mean()

print(workingday_hour_rental.loc[
    [(0, 8), (0, 13), (0, 17),
     (1, 8), (1, 13), (1, 17)]
])
```

```python
holiday_rental = df.groupby("holiday")["cnt"].mean()
print(f"holiday rental : {holiday_rental}")
```

```python
temp_rental = df.groupby(
    pd.cut(df["temp"], bins=5)
)["cnt"].mean()
print(f"temp rental : {temp_rental}")
```

```python
humidity_rental = df.groupby(
    pd.cut(df["hum"], bins=5)
)["cnt"].mean()

print(f"humidity rental : {humidity_rental}")
```

```python
temp_hum_count = df.pivot_table(
    index=pd.cut(df["temp"], bins=5),
    columns=pd.cut(df["hum"], bins=5),
    values="cnt",
    aggfunc="count"
)

print(f"temp humidity count : {temp_hum_count}")
```

```python
corr_df = df.drop(columns=["instant", "casual", "registered"])
correlation = corr_df.corr(numeric_only=True)
print(correlation["cnt"].sort_values(ascending=False))
```

Higher temperatures are associated with higher rental demand, while higher humidity generally shows lower demand; the combined pivot and correlation analysis help identify feature relationships and guide ML feature selection.
